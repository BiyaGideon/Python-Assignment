```python
print("===== SIMPLE CALCULATOR =====")

num1 = float(input("Enter first number: "))
op = input("Enter operator (+, -, , /, //, %, *): ")
num2 = float(input("Enter second number: "))

if op == "+":
    result = num1 + num2

elif op == "-":
    result = num1 - num2

elif op == "*":
    result = num1 * num2

elif op == "/":
    if num2 == 0:
        print("Cannot divide by zero")
        result = 0
    else:
        result = num1 / num2

elif op == "//":
    if num2 == 0:
        print("Cannot divide by zero")
        result = 0
    else:
        result = num1 // num2

elif op == "%":
    if num2 == 0:
        print("Cannot divide by zero")
        result = 0
    else:
        result = num1 % num2

elif op == "**":
    result = num1 ** num2

else:
    print("Wrong operator")
    result = 0

print("Result:", result)
```
