```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))
gpa = float(input("Enter your GPA: "))
courses = input("Enter your courses (comma separated): ")

courses_list = courses.split(",")

num_courses = len(courses_list)

if gpa >= 3.5:
    deans = "Yes"
else:
    deans = "No"

if age < 18:
    status = "Minor"
else:
    status = "Adult"

print("\n===== STUDENT PROFILE =====")
print("Name:", name)
print("Age:", age, "(", status, ")")
print("GPA:", gpa)
print("Courses:", courses_list)
print("Number of Courses:", num_courses)
print("Dean's List:", deans)
```
